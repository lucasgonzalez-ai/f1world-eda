## 7. ¿Cómo fue el rendimiento de los pilotos argentinos con el paso de los años? ¿Cuál fue el año con mayor proporción de victorias?

```python
# Obtener lista de Ids de los argentinos
driver_ids_argentinos = df_drivers[df_drivers['nationality'] == 'Argentinian']['driverId'].unique()
driver_ids_argentinos
```

```python
# Seleccionamos las filas en df_results donde los 'driverId' coinciden con los argentinos
resultados_argentinos = df_results[df_results['driverId'].isin(driver_ids_argentinos)].copy()
```

```python
# Hacemos un merge para obtener el año de cada carrera, usando raceId como columna comun
resultados_argentinos = pd.merge(resultados_argentinos, df_races[['raceId', 'year']], on='raceId', how='left')
resultados_argentinos
```

```python
# Filtramos las filas donde la posición es 1 (victoria)
victorias_argentinos = resultados_argentinos[resultados_argentinos['positionOrder'] == 1]

# Contamos cuántas victorias hubo por año
victorias_por_año = victorias_argentinos.groupby('year').size().reset_index(name='victorias_argentinas')
victorias_por_año
```

```python
# Calculamos cantidad total de carreras por año (todas, no solo de argentinos) ---
# Usamos df_races para contar la cantidad de carreras por año
total_carreras_por_año = df_races.groupby('year').size().reset_index(name='total_carreras')
total_carreras_por_año
```

```python
# Unimos los dataframes de victorias y total de carreras
# Hacemos un merge para tener las victorias y el total de carreras en un solo dataframe
# Usamos 'how=right' para asegurarnos de incluir todos los años de las carreras,
# incluso si no hubo victorias argentinas en ese año (fillna(0) los convierte a cero).
proporcion_victorias = pd.merge(victorias_por_año, total_carreras_por_año, on='year', how='right').fillna(0)
proporcion_victorias
```

```python
# Calcular la proporción de victorias por año
proporcion_victorias['proporcion'] = proporcion_victorias['victorias_argentinas'] / proporcion_victorias['total_carreras']
```

```python
# Determinar el año con mayor proporción
# Usamos .idxmax() para encontrar el índice (que en este caso es el año)
# del valor máximo en la columna 'proporcion'
año_con_mayor_proporcion = proporcion_victorias.loc[proporcion_victorias['proporcion'].idxmax()]
```

## Graficamos con matplotlib para responder la pregunta 7

```python
plt.figure(figsize=(12, 7))
plt.plot(proporcion_victorias['year'], proporcion_victorias['proporcion'], marker='o', linestyle='-', color='b')
plt.title('Proporción de victorias argentinas por año', fontsize=16)
plt.xlabel('Año', fontsize=12)
plt.ylabel('Proporción de victorias', fontsize=12)
plt.grid(True)
plt.show()
```

## 8. ¿Es posible sacar conclusiones sobre el rendimiento de Franco Colapinto?

```python
df_qualifying = pd.read_csv('/qualifying.csv')
df_constructor_results = pd.read_csv('/constructor_results.csv')
df_constructors = pd.read_csv('/constructors.csv')
```

```python
# Identificar el Id de Franco Colapinto
colapinto_id = df_drivers[(df_drivers['forename'] == 'Franco') & (df_drivers['surname'] == 'Colapinto')]['driverId'].iloc[0]
colapinto_id
```

```python
df_drivers.loc[df_drivers['driverId'] == 861]
```

```python
# Obtengo los resultados de carrera de Colapinto
resultados_colapinto = df_results[df_results['driverId'] == colapinto_id].copy()
```

```python
# Conecto los resultados con los nombres de los circuitos y años de las carreras
resultados_colapinto = pd.merge(resultados_colapinto, df_races[['raceId', 'year', 'name']], on='raceId', how='left')
resultados_colapinto
```

```python
# Identifico al compañero de equipo de Colapinto para cada carrera
# Unimos los resultados de Colapinto con los resultados de todos los pilotos en la misma carrera
# La clave es el 'constructorId' y 'raceId'
# El sufijo _colapinto marca las columnas que vienen de su propio registro.
# El sufijo _compañero marca las columnas que vienen de la tabla con todos los pilotos.

resultados_equipo = pd.merge(resultados_colapinto[['raceId', 'constructorId', 'positionOrder', 'driverId']],
                             df_results[['raceId', 'constructorId', 'positionOrder', 'driverId']],
                             on=['raceId', 'constructorId'],
                             how='left',
                             suffixes=('_colapinto', '_compañero'))
```

```python
# Elimino filas donde el compañero es el mismo Colapinto
resultados_equipo = resultados_equipo[resultados_equipo['driverId_colapinto'] != resultados_equipo['driverId_compañero']]
resultados_equipo
```

```python
# Cuento las carreras en las que Colapinto terminó por delante
# Nos basamos en la 'positionOrder' (la menor posición es la mejor)
colapinto_adelante_carrera = (resultados_equipo['positionOrder_colapinto'] < resultados_equipo['positionOrder_compañero']).sum()
total_carreras_comparables = len(resultados_equipo)
total_carreras_comparables
```

```python
if total_carreras_comparables > 0:
    proporcion_adelante = (colapinto_adelante_carrera / total_carreras_comparables) * 100
    print(f"En las {total_carreras_comparables} carreras comparables, Franco Colapinto terminó por delante de su compañero de equipo en {colapinto_adelante_carrera} ocasiones.")
    print(f"Esto representa un {proporcion_adelante:.2f}% de las veces.\n")
else:
    print("No se encontraron suficientes datos para comparar a Colapinto con su compañero de equipo.\n")
```

```python
# Muestro resultados de carrera de forma cronológica
print("Evolución de la posición final de Franco Colapinto:")
print(resultados_colapinto[['year', 'name', 'positionOrder']].sort_values(by=['year', 'name']))
```

```python
# Analizo la posición de clasificación (si está disponible)
# Unimos los resultados de calificación con los de Colapinto
qualifying_colapinto = df_qualifying[df_qualifying['driverId'] == colapinto_id].copy()
qualifying_colapinto = pd.merge(qualifying_colapinto, df_races[['raceId', 'year', 'name']], on='raceId', how='left')

print("\nEvolución de la posición de clasificación de Franco Colapinto:")
# 'position' aquí se refiere a la posición antes de comenzar la carrera
print(qualifying_colapinto[['year', 'name', 'position']].sort_values(by=['year', 'name']))
```

```python
# Visualización de la evolución de Colapinto

if not resultados_colapinto.empty:
    plt.figure(figsize=(12, 6))
    plt.plot(resultados_colapinto['year'].astype(str) + '-' + resultados_colapinto['name'],
             resultados_colapinto['positionOrder'],
             marker='o', linestyle='-', color='g')
    plt.gca().invert_yaxis() # Invertimos el eje Y para que la posición 1 esté arriba
    plt.title('Evolución de la posición final de Franco Colapinto')
    plt.xlabel('Carrera')
    plt.ylabel('Posición Final')
    plt.xticks(rotation=90)
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```
